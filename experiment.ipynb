```python
# ==========================
# Import Libraries
# ==========================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import cv2
from sklearn.preprocessing import LabelEncoder

# ==========================
# Introduction
# ==========================
print("=" * 60)
print("INTRODUCTION TO PYTHON LIBRARIES FOR AI")
print("=" * 60)

# ==========================
# NumPy
# ==========================
print("\n----- NumPy -----")

numbers = np.array([10, 20, 30, 40, 50])

print("Array:", numbers)
print("Sum:", np.sum(numbers))
print("Mean:", np.mean(numbers))
print("Maximum:", np.max(numbers))
print("Minimum:", np.min(numbers))

# ==========================
# Pandas
# ==========================
print("\n----- Pandas -----")

student = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(student)

print("\nDataFrame:")
print(df)

# ==========================
# Matplotlib
# ==========================
print("\n----- Matplotlib -----")
print("Displaying Line Chart....")

plt.figure(figsize=(6, 4))
plt.plot(df["Name"], df["Marks"], marker="o")
plt.title("Student Marks")
plt.xlabel("Students Name")
plt.ylabel("Marks")

plt.show()

# ==========================
# Scikit-learn
# ==========================
print("\n----- Scikit-learn -----")

encoder = LabelEncoder()

# Encode student names into numbers
df["Student ID"] = encoder.fit_transform(df["Name"])

print("After Encoding Names into Numbers:")
print(df)

# ==========================
# TensorFlow
# ==========================
print("\n----- TensorFlow -----")

tensor1 = tf.constant([1, 2, 3])
tensor2 = tf.constant([4, 5, 6])

print("Tensor1:", tensor1.numpy())
print("Tensor2:", tensor2.numpy())

result = tf.add(tensor1, tensor2)

print("Tensor Addition:", result.numpy())

# ==========================
# OpenCV
# ==========================
print("\n----- OpenCV -----")

# Create a blank image
image = np.zeros((200, 200, 3), dtype=np.uint8)

# Draw a green rectangle
cv2.rectangle(image, (50, 50), (150, 150), (0, 255, 0), 3)

# Display using Matplotlib
plt.figure(figsize=(4, 4))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("OpenCV Shape")
plt.axis("off")
plt.show()

# ==========================
# End of Program
# ==========================
print("\n" + "=" * 60)
print("Experiment Complete")
print("=" * 60)
```
